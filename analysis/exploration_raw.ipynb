```python
products = pd.read_csv(
    os.path.join(path, "product.csv")
)
print(products.shape)
```

```python
demographics = pd.read_csv(
    os.path.join(path, "hh_demographic.csv")
)

print(demographics.shape)
```

```python
demographics
```

```python
df = (
    transactions
    .merge(
        products,
        on="PRODUCT_ID",
        how="left",
        validate="m:1"
    )
    .merge(
        demographics,
        on="household_key",
        how="left",
        validate="m:1" 
    )
)
```

```python
df
```

```python
past = df[df["WEEK_NO"] <= 52].copy()
future = df[df["WEEK_NO"] > 52].copy()
```

```python
print("Overall weeks:", df["WEEK_NO"].min(), "to", df["WEEK_NO"].max())

print("\nPAST")
print("Weeks:", past["WEEK_NO"].min(), "to", past["WEEK_NO"].max())
print("Households:", past["household_key"].nunique())
print("Baskets:", past["BASKET_ID"].nunique())

print("\nFUTURE")
print("Weeks:", future["WEEK_NO"].min(), "to", future["WEEK_NO"].max())
print("Households:", future["household_key"].nunique())
print("Baskets:", future["BASKET_ID"].nunique())
```

```python
past_households = set(past["household_key"])
future_households = set(future["household_key"])

print("In both:", len(past_households & future_households))
print("Past only:", len(past_households - future_households))
print("Future only:", len(future_households - past_households))
```

```python
past_behavior = (
    past.groupby("household_key")
    .agg(
        trips=("BASKET_ID", "nunique"),
        spend=("SALES_VALUE", "sum"),
        active_weeks=("WEEK_NO", "nunique"),
        commodity_breadth=("COMMODITY_DESC", "nunique"),
        product_breadth=("PRODUCT_ID", "nunique")
    )
    .reset_index()
)
```

```python
past_behavior["trips_per_active_week"] = (
    past_behavior["trips"] / past_behavior["active_weeks"]
)

past_behavior["spend_per_trip"] = (
    past_behavior["spend"] / past_behavior["trips"]
)
```

```python
future_value = (
    future.groupby("household_key")
    .agg(
        future_trips=("BASKET_ID", "nunique"),
        future_spend=("SALES_VALUE", "sum"),
        future_active_weeks=("WEEK_NO", "nunique")
    )
    .reset_index()
)
```

```python
analysis = past_behavior.merge(
    future_value,
    on="household_key",
    how="left",
    validate="1:1"
)

analysis[
    ["future_trips", "future_spend", "future_active_weeks"]
] = analysis[
    ["future_trips", "future_spend", "future_active_weeks"]
].fillna(0)
```

```python
analysis.describe().T
```

```python
analysis["high_value"] = (
    analysis["future_trips"] >= 78
).astype(int)
```

```python
analysis["high_value"].value_counts()
```

```python
analysis["high_value"].mean()
```

```python
analysis["breadth_quartile"] = pd.qcut(
    analysis["commodity_breadth"],
    4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)

breadth_summary = (
    analysis.groupby("breadth_quartile", observed=True)
    .agg(
        households=("household_key", "count"),
        avg_past_trips=("trips", "mean"),
        avg_future_trips=("future_trips", "mean"),
        avg_future_spend=("future_spend", "mean"),
        high_value_rate=("high_value", "mean")
    )
)

breadth_summary
```

```python
import statsmodels.formula.api as smf

model = smf.logit(
    "high_value ~ trips + commodity_breadth",
    data=analysis
).fit()

print(model.summary())
```

```python
analysis["trips_10"] = analysis["trips"] / 10
analysis["commodity_breadth_10"] = analysis["commodity_breadth"] / 10

model = smf.logit(
    "high_value ~ trips_10 + commodity_breadth_10",
    data=analysis
).fit()

print(model.summary())
```

```python
analysis["trip_quartile"] = pd.qcut(
    analysis["trips"],
    4,
    labels=["Q1", "Q2", "Q3", "Q4"]
)
```

```python
analysis["high_breadth_within_trip_group"] = (
    analysis.groupby("trip_quartile", observed=True)["commodity_breadth"]
    .transform(lambda x: x >= x.median())
)
```

```python
comparison = (
    analysis.groupby(
        ["trip_quartile", "high_breadth_within_trip_group"],
        observed=True
    )
    .agg(
        households=("household_key", "count"),
        avg_past_trips=("trips", "mean"),
        avg_commodity_breadth=("commodity_breadth", "mean"),
        avg_future_trips=("future_trips", "mean"),
        high_value_rate=("high_value", "mean")
    )
)

comparison
```

```python
import numpy as np
import statsmodels.formula.api as smf

analysis["log_trips"] = np.log1p(analysis["trips"])
analysis["log_future_trips"] = np.log1p(analysis["future_trips"])
analysis["commodity_breadth_10"] = analysis["commodity_breadth"] / 10

model_trips = smf.ols(
    "log_future_trips ~ log_trips + commodity_breadth_10",
    data=analysis
).fit(cov_type="HC3")

print(model_trips.summary())
```

```python
analysis["log_spend"] = np.log1p(analysis["spend"])

model_trips_2 = smf.ols(
    "log_future_trips ~ log_trips + log_spend + commodity_breadth_10",
    data=analysis
).fit(cov_type="HC3")

print(model_trips_2.summary())
```
